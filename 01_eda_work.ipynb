```python
import pandas as pd

attributes  = pd.read_csv('data/features_attributes.csv')
financials  = pd.read_csv('data/features_financials.csv')
clickstream = pd.read_csv('data/feature_clickstream.csv')
loans       = pd.read_csv('data/lms_loan_daily.csv')

for name, df in [('attributes', attributes), ('financials', financials),
                 ('clickstream', clickstream), ('loans', loans)]:
    print(f'{name:12s} {df.shape[0]:>7,} rows  {df.shape[1]:>3} cols  '
          f'{df.Customer_ID.nunique():>6,} customers')
```

```python
print('unique loans     :', loans.loan_id.nunique())
print('unique customers :', loans.Customer_ID.nunique())
print('tenure values    :', loans.tenure.unique())
print()
print('rows per loan:')
print(loans.groupby('loan_id').size().value_counts())
print()
print(loans.head(12).to_string())
```

```python
arrears = loans.groupby('installment_num').apply(
    lambda g: (g.overdue_amt > 0).mean(), include_groups=False)

arrears.round(3).to_frame('pct_with_overdue')
```

```python
import numpy as np

l = loans.copy()
l['snapshot_date'] = pd.to_datetime(l.snapshot_date)

# overdue balance expressed in whole instalments
l['installments_missed'] = np.ceil(
    l.overdue_amt / l.due_amt.replace(0, np.nan)).fillna(0).astype(int)

# walk back one month per missed instalment → when arrears began
l['first_missed'] = l.apply(
    lambda r: r.snapshot_date - pd.DateOffset(months=int(r.installments_missed))
    if r.installments_missed > 0 else pd.NaT, axis=1)

l['dpd'] = np.where(l.overdue_amt > 0,
                    (l.snapshot_date - l.first_missed).dt.days, 0)

pd.DataFrame([
    {'mob': m,
     'dpd>0':  round((l[l.installment_num == m].dpd > 0).mean(), 4),
     'dpd>30': round((l[l.installment_num == m].dpd > 30).mean(), 4),
     'dpd>60': round((l[l.installment_num == m].dpd > 60).mean(), 4)}
    for m in [3, 5, 6, 9]
])
```

```python
apps = loans[loans.installment_num == 0][['loan_id','Customer_ID','loan_start_date']]

for name, df in [('attributes', attributes), ('financials', financials)]:
    chk = apps.merge(df[['Customer_ID','snapshot_date']], on='Customer_ID')
    match = (chk.snapshot_date == chk.loan_start_date).mean()
    print(f'{name:12s} snapshot_date == loan_start_date : {match:.1%}')

print()
print('clickstream months per customer:')
print(clickstream.groupby('Customer_ID').size().value_counts())
print()
print(f'applicants with any clickstream: '
      f'{apps.Customer_ID.isin(clickstream.Customer_ID).mean():.1%}')
```

```python
cs = clickstream.copy()
cs['snapshot_date'] = pd.to_datetime(cs.snapshot_date)
a = apps.copy()
a['loan_start_date'] = pd.to_datetime(a.loan_start_date)

j = cs.merge(a, on='Customer_ID')
print(f'total clickstream rows joined to an application : {len(j):,}')
print(f'rows dated BEFORE the application (usable)      : {(j.snapshot_date < j.loan_start_date).sum():,}')
print(f'rows dated ON or AFTER (leakage if used)        : {(j.snapshot_date >= j.loan_start_date).sum():,}')
```

```python
fin = pd.read_csv('data/features_financials.csv', dtype=str)
att = pd.read_csv('data/features_attributes.csv', dtype=str)

rows = []
for col in ['Annual_Income','Num_of_Loan','Num_of_Delayed_Payment',
            'Changed_Credit_Limit','Outstanding_Debt',
            'Amount_invested_monthly','Monthly_Balance']:
    bad = fin[col][~fin[col].str.fullmatch(r'-?\d*\.?\d+', na=True)]
    if len(bad):
        rows.append({'column': col, 'bad_rows': len(bad),
                     'example': bad.iloc[0]})

rows += [
 {'column':'SSN','bad_rows':(~att.SSN.str.fullmatch(r'\d{3}-\d{2}-\d{4}',na=False)).sum(),'example':att.SSN[~att.SSN.str.fullmatch(r'\d{3}-\d{2}-\d{4}',na=False)].iloc[0]},
 {'column':'Occupation','bad_rows':(att.Occupation=='_______').sum(),'example':'_______'},
 {'column':'Credit_Mix','bad_rows':(fin.Credit_Mix=='_').sum(),'example':'_'},
 {'column':'Payment_Behaviour','bad_rows':(fin.Payment_Behaviour=='!@9#%8').sum(),'example':'!@9#%8'},
]

pd.DataFrame(rows).sort_values('bad_rows', ascending=False)
```
